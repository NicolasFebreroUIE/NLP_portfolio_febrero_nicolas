### Exercise 02: Text to Speech

```python
import pyttsx3

engine = pyttsx3.init()
engine.setProperty('rate', 150) 
voices = engine.getProperty('voices')
for voice in voices:
    if "EN" in voice.id.upper() or "ENGLISH" in voice.name.upper():
        engine.setProperty('voice', voice.id)
        break

text = "Hello my super class from Vigo, beautiful day we are having!"
engine.save_to_file(text, 'local_exercise_2.wav')

engine.runAndWait()
print("Audio saved successfully as 'local_exercise_2.wav'")
```
