### Exercise 05: Beyond STT and TTS

```python
from textblob import TextBlob

# best for whisper usage, so it worked well on my code
text_from_speech = "Hello my super class from Vigo, beautiful day we are having!"
blob = TextBlob(text_from_speech)
print(f"Text: {text_from_speech}")
print(f"Polarity score: {blob.sentiment.polarity}") 

# Just a loop wit hthree simple sentiments.
if blob.sentiment.polarity > 0:
    print("Result: POSITIVE sentiment")
elif blob.sentiment.polarity < 0:
    print("Result: NEGATIVE sentiment")
else:
    print("Result: NEUTRAL sentiment")
```
