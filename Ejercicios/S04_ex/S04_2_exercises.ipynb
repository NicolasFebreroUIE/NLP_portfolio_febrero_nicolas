### Exercise 1 (Easy)
Generate all bigrams and trigrams from a sentence.

```python
sentence = "I love natural language processing"

def get_ngrams(tokens, n):
    ngrams = []
    for i in range(len(tokens) - n+1):
        ngram = tuple(tokens)
        ngrams.append(ngram)
    return ngrams

tokens = sentence.lower().split()
# Get bigrams and trigrams
bigrams = get_ngrams(tokens, 2)
trigrams = get_ngrams(tokens, 3)

print(f"sentence: {sentence}")
print(f"Bigrams: {bigrams}")
print(f"Trigrams: {trigrams}")
```

### Exercise 2 (Easy)
Calculate the probability of a bigram using Maximum Likelihood Estimation (MLE).

```python
corpus = ["i love nlp", "i love python", "i hate bugs", "nlp is great"]

# Calculate P(love | i) using MLE
# P(love | i) = count(i, love) / count(i)

# part 01 
all_w = []
pairs = []
for frase in corpus:
    words = frase.lower().split()
    all_w.extend(words)
    
    
    for i in range(len(words) - 1):
        pairs.append((words[i], words[i+1]))

Count_unigrams = Counter(all_w)
Count_bigrams = Counter(pairs)

# part 02

count_i = Count_unigrams['i']
count_i_love = Count_bigrams[('i', 'love')]

prob = count_i_love / count_i
print(f"Prob P(love | i) = {prob}")

```
